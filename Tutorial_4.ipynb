Question 1: Is this data in tidy format?
Ans:This Data is not in tidy format.

Queston 2:If not, explain which tidy-data rule is violated.
Ans:In tidy format,each variable forms a column and each observation forms a row.Here in this data format, "Math","Science","English" are displayed as variable.They are values of variable "Subject".Each row displays three observations,instead of one.

```python
#3.Using Python/pandas, reshape the dataset so that it has the columns:Student,Subject,Score

import pandas as pd
df=pd.DataFrame({
    "Student":["Ali", "Sara", "John"],
    "Math":[75, 88, 65],
    "Science":[80, 91, 70],
    "English":[72, 85, 68]
})
tidy = df.melt(id_vars="Student", var_name="Subject", value_name="Score")
tidy = tidy.sort_values("Student").reset_index(drop=True)  # optional
print(tidy)

```

Question 2:
a)What is the tidy-data problem with this dataset?

Ans: q1_sales,q2_sales,q3_sales are dsiplayed as variable in column.But they are values of sales Column.

```python
#Question 2:b) Use pandas to convert it into:
#Country    Quarter    Sales
#Malaysia   Q1         12000
#Malaysia   Q2         15000
import pandas as pd

sales = pd.DataFrame({
    "Country":  ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000]
})

tidy_sales = sales.melt(id_vars="Country", var_name="Quarter", value_name="Sales")
tidy_sales["Quarter"] = tidy_sales["Quarter"].str.replace("_Sales", "", regex=False)
tidy_sales = tidy_sales.sort_values(["Country", "Quarter"]).reset_index(drop=True)
print(tidy_sales)
```

```python
#Question 2
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
}
df = pd.DataFrame(data)

print("Average study hours:", df["Study_Hours"].mean())
print("Average score:", df["Score"].mean())
print("Highest score:", df.loc[df["Score"].idxmax(), "Student"])
print("Studied the most:", df.loc[df["Study_Hours"].idxmax(), "Student"])
print("Correlation:", df["Study_Hours"].corr(df["Score"]))


plt.scatter(df["Study_Hours"], df["Score"])
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.grid(0.3)
plt.show()
```
